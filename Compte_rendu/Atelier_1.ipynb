## 1. API REST : comment lire la réponse ?

- http_200 : Cela indique que la requete à bien été traitée par le serveur
- pagination : Total : c'est le nombre total d'éléments disponibles, limit : c'est le nombre d'éléments renvoyés par la requête, skip : c'est le nombre d'éléments ignorés avant de commencer à renvoyer les résultats
- api_vs_csv : Une API permet de récupérer les données alors que le csv est un format de fichier
- date_collecte : Il faut la conserver pour savoir quand les données sont collectées
